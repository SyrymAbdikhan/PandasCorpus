## Looking up a notebook and resolving its GitHub URL

Each notebook is identified by a hash. Use `link_hash_map.csv` to resolve a hash back to its original GitHub file URL

```python
link_hash_map = pd.read_csv('data/link_hash_map.csv')

link_hash_map.head(3)
```

```python
# Pick the first notebook hash from code_cells
target_hash = code_cells['notebook_hash'].iloc[0]
print('Target hash:', target_hash)

# Retrieve its code cells
nb_cells = code_cells[code_cells['notebook_hash'] == target_hash].sort_values('cell_id')
print(f'Found {len(nb_cells)} code cells')
nb_cells.head()
```

```python
# Resolve the hash to its GitHub file URL
match = link_hash_map[link_hash_map['hash'] == target_hash]
if not match.empty:
    print('Repo name:', match['repo_name'].values[0])
    print('Repo URL:', match['repo_url'].values[0])
    print('File URL:', match['file_url'].values[0])
else:
    print('Hash not found in link_hash_map')
```

## Filtering notebooks that use a specific operation

Find all code cells that call `groupby`

```python
operation = 'groupby'

mask = code_cells['source'].str.contains(r'\.groupby\s*\(', regex=True, na=False)
groupby_cells = code_cells[mask]

print(f'Cells containing .{operation}(): {len(groupby_cells):,}')
print(f'Unique notebooks: {groupby_cells["notebook_hash"].nunique():,}')
```

```python
# Print the source of the first 3 matching cells
for _, row in groupby_cells.head(3).iterrows():
    print(f'--- notebook {row["notebook_hash"]} cell {row["cell_id"]} ---')
    print(row['source'][:300])
    print()
```
